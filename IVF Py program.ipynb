ANALYSIS FOR PATIENT INFORMATION IN PANDAS

```python
print("Average age of patients:")
print(a['Age'].mean(), "\n")

print("Median age of patients:")
print(a['Age'].median(), "\n")

print("Distribution of Marital Status (counts):")
print(a['Marital Status'].value_counts(), "\n")

print("Unique Marital Status categories:")
print(a['Marital Status'].unique(), "\n")

print("Primary vs Secondary infertility (counts):")
print(a['Type of Infertility'].value_counts(), "\n")

print("Most common infertility type:")
print(a['Type of Infertility'].mode()[0], "\n")

print("Season with the highest number of IVF Cycles (counts):")
print(a['Season'].value_counts(), "\n")

print("Most frequent season for IVF cycles:")
print(a['Season'].mode()[0], "\n")
```

```python
hormone_columns = ['TSH (µIU/mL)', 'Prolactin (ng/mL)',
                'AMH Level (ng/mL)', 'LH (mIU/mL)', 'FSH (mIU/mL)']

print("Avg TSH, Prolactin, AMH, LH, FSH values:")
print(a[hormone_columns].mean(), "\n")

print("No of Patients with Childhood Diseases (Mumps, Measles, Chickenpox, None):")
print(a['Childish Diseases'].value_counts(dropna=True), "\n")

print("No of Patients with Childhood Diseases (Mumps, Measles, Chickenpox, None):")
print(a['Childish Diseases'].value_counts(dropna=False), "\n")

print("Distribution of Semen Analysis Results:")
print(a['Semen Analysis'].value_counts(), "\n")
```

```python
print("Percentage of IVF Cycles With Positive Outcome:?")
res_counts = a['Result'].value_counts() 
print(res_counts)
pos_count = res_counts['Positive'] 
total = len(a)
pos_rate = (pos_count / total) * 100
print(f"Positive Rate: {pos_rate:.2f}%  \n")

print("Which Stimulation Protocol has the Worst Outcome:?")
a['Success'] = (a['Result'] == 'Negative').astype(int) 
prot_suc = a.groupby('Stimulation Protocol')['Success'].mean() 
prot_suc = prot_suc.sort_values(ascending=True)  
print(prot_suc, "\n")

print("Which Stimulation Protocol has the Best Outcome:?")
a['Success'] = (a['Result'] == 'Positive').astype(int) 
prot_suc = a.groupby('Stimulation Protocol')['Success'].mean() 
prot_suc = prot_suc.sort_values(ascending=False)  
print(prot_suc, "\n")

print("Which Medication Results in the Highest Average Embryo Formation:?")
medi_embryos = a.groupby('Medications Used')['Embryos Formed'].mean()  
medi_embryos = medi_embryos.sort_values(ascending=False) 
print(medi_embryos, "\n")
```

ANALYSIS FOR PATIENT INFORMATION IN MATPLOT.LIB

```python
print("MATPLOTLIB LINE PLOTS")

print("Plot the number of IVF cycles conducted each year using a line graph:")
cycles_per_year = a["Year"].value_counts().sort_index()
plt.plot(cycles_per_year.index, cycles_per_year.values, color='yellow', marker='o')
plt.title("IVF Cycles per Year-LinePlot")
plt.xlabel("Year")
plt.ylabel("Cycles")
plt.grid(True)
plt.show()

a["Success"] = (a["Result"] == "Positive").astype(int)
print("Plot the IVF success rate for each year as a line graph:")
success_rate_year = a.groupby("Year")["Success"].mean() 
plt.figure(facecolor='pink')
plt.plot(success_rate_year.index, success_rate_year.values, marker='o', color='green')
plt.title("Success Rate per Year-LinePlot")
plt.xlabel("Year")
plt.ylabel("Success Rate")
plt.grid(True)
plt.show()

print("MATPLOTLIB BAR CHARTS")

print("Create a bar chart showing the average number of eggs retrieved for each stimulation protocol:")
protocol_eggs = a.groupby("Stimulation Protocol")["Eggs Retrieved"].mean()
plt.figure(facecolor='lightblue')
plt.bar(protocol_eggs.index, protocol_eggs.values, color='skyblue', edgecolor='black')
plt.title("Eggs Retrieved per Protocol-BarChart")
plt.ylabel("Eggs Retrieved")
plt.xticks(rotation=45)
plt.show()

print("Create a bar chart showing the average number of embryos formed for each medication used:")
med_embryos = a.groupby("Medications Used")["Embryos Formed"].mean()
plt.figure(facecolor='pink')
plt.bar(med_embryos.index, med_embryos.values, color='orange', edgecolor='black')
plt.title("Embryos Formed per Medication-BarChart")
plt.ylabel("Embryos Formed")
plt.xticks(rotation=45)
plt.show()

print("MATPLOTLIB PIE CHARTS")

print("Plot a pie chart showing the percentage distribution of IVF results (Pos & Neg):")
result_counts = a["Result"].value_counts()
colors = ['red', 'blue']
plt.figure(facecolor='lightblue')
plt.pie(result_counts.values,labels=result_counts.index,colors=colors,autopct="%1.1f%%")
plt.title("Success vs Failure-PieChart")
plt.show()

print("Plot a pie chart showing the percentage distribution of Childish Diseases:")
disease_counts = a["Childish Diseases"].value_counts()
plt.figure(facecolor='pink')
plt.pie(disease_counts.values, labels=disease_counts.index, autopct="%1.1f%%")
plt.title("Childish Diseases-PieChart")
plt.show()

print("Plot a pie chart showing the percentage of Primary vs Secondary infertility case:")
t_counts = a["Type of Infertility"].value_counts()
colors = ['green', 'pink']
plt.figure(facecolor='lightblue')
plt.pie(t_counts.values, labels=t_counts.index, autopct="%1.1f%%",colors=colors)
plt.title("Primary vs Secondary Infertility-PieChart")
plt.show()
```

ANALYSIS FOR PATIENT INFORMATION IN SEABORN 

```python
print("SEABORN DISTRIBUTION PLOTS")
print("What does the kde=True argument add to a histogram, and when is it useful:")
sns.histplot(a["Age"], kde=True,color='yellow',edgecolor='blue')
plt.title("Age Distribution-Histplot")
plt.show()

print("Why might you choose different edgecolor values for histograms:")
sns.histplot(a["Eggs Retrieved"], kde=True,color='blue',edgecolor='red')
plt.title("Eggs Retrieved Distribution-Histplot")
plt.show()

print("How can you interpret the shape of the Age Distribution plot in terms of skewness:")
sns.histplot(a["Prolactin (ng/mL)"], kde=True,color='black',edgecolor='orange')
plt.title("Prolactin Distribution-Histplot")
plt.show()

print("If the TSH Distribution shows outliers, what preprocessing steps could you take before modeling:")
sns.histplot(a["TSH (µIU/mL)"], kde=True,color='yellow',edgecolor='black')
plt.title("TSH Distribution-Histplot")
plt.show()

print("SEABORN COUNT PLOTS")
print("What does sns.countplot visualize compared to sns.histplot:")
sns.countplot(x="Type of Infertility", data=a,color='indigo',edgecolor='pink')
plt.title("Infertility Type Count-Histplot")
plt.show()

print("Why might rotating x-axis labels (plt.xticks(rotation=45)) be necessary:")
sns.countplot(x="Diagnosis", data=a,color='purple',edgecolor='orange')
plt.title("Diagnosis Count-Histplot")
plt.xticks(rotation=45)
plt.show()

print("Why might rotating x-axis labels (plt.xticks(rotation=90)) be necessary:")
sns.countplot(x="Stimulation Protocol", data=a,color='pink',edgecolor='blue')
plt.title("Stimulation Protocol Count-Histplot")
plt.xticks(rotation=90)
plt.show()

print("If the Diagnosis Count plot shows imbalance, how could that affect downstream machine learning models:")
sns.countplot(x="Result", data=a,color='black',edgecolor='orange')
plt.title("Result Count-Histplot")
plt.show()

print("How would you modify the Season Count plot to show percentages instead of raw counts:")
sns.countplot(x="Season", data=a,color='brown',edgecolor='orange')
plt.title("Season Count-Histplot")
plt.show()

print("SEABORN RELATIONSHIP PLOTS")
print("In the Age vs Eggs Retrieved scatterplot, what does the hue = Result parameter achieve:")
sns.scatterplot(x="Age", y="Eggs Retrieved", hue="Result", data=a)
plt.title("Age vs Eggs Retrieved-Scatterplot")
plt.show()

print("How can you detect correlation patterns between AMH Level and Embryos Formed using scatterplots:")
sns.scatterplot(x="AMH Level (ng/mL)", y="Embryos Formed", hue="Result", data=a)
plt.title("AMH vs Embryos Formed-Scatterplots")
plt.show()

print("If the Diagnosis vs Embryos Formed boxplot shows overlapping distributions, what statistical test could you apply:")
sns.boxplot(x="Diagnosis", y="Embryos Formed", data=a)
plt.title("Diagnosis vs Embryos Formed-Boxplot")
plt.xticks(rotation=45)
plt.show()

print("If the Medications Used formed barplot using mean() test could you apply:")
med_success = a.groupby("Medications Used")["Success"].mean().reset_index()
sns.barplot(x="Medications Used", y="Success", data=med_success,color='blue',edgecolor='black')
plt.title("Medication vs Success Rate-Barplot")
plt.xticks(rotation=45)
plt.show()

print("- What are the limitations of correlation analysis in medical datasets:")
hormone_columns = ["TSH (µIU/mL)", "Prolactin (ng/mL)",
                "AMH Level (ng/mL)", "LH (mIU/mL)", "FSH (mIU/mL)"]
corr_data = a[hormone_columns + ["Success"]].corr()
sns.heatmap(corr_data, annot=True,color='yellow',edgecolor='black')
plt.title("Correlation: Hormones vs Outcome - HaetMap")
plt.show()
```
